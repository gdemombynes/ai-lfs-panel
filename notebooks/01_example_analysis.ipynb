# Example analysis

A minimal walkthrough of the intended workflow: load, explore, process, save.

## 1. Setup

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from analysis.utils import load_data, save_data

plt.rcParams["figure.figsize"] = (10, 5)
```

## 2. Load data

Replace the synthetic frame below with `load_data("../data/raw/your_file.csv")`.

```python
rng = np.random.default_rng(0)
df = pd.DataFrame(
    {
        "date": pd.date_range("2024-01-01", periods=100),
        "value": rng.standard_normal(100).cumsum(),
        "category": rng.choice(["A", "B", "C"], size=100),
    }
)
df.head()
```

## 3. Explore

```python
df.describe()
```

```python
fig, (ax1, ax2) = plt.subplots(1, 2)
ax1.plot(df["date"], df["value"])
ax1.set_title("Value over time")
df.groupby("category")["value"].mean().plot(kind="bar", ax=ax2)
ax2.set_title("Mean value by category")
fig.tight_layout()
```

## 4. Process

```python
processed = df.assign(rolling_mean=df["value"].rolling(7).mean())
processed.tail()
```

## 5. Save

Outputs go to `data/processed/`, which git ignores.

```python
save_data(processed, "../data/processed/example_processed.csv")
load_data("../data/processed/example_processed.csv").shape
```
